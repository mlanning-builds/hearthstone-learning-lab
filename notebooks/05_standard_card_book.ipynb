```python
from pathlib import Path
import sys, json
ROOT = next((p for p in (Path.cwd(), *Path.cwd().parents) if (p / "standard" / "catalog.py").exists()), None)
if ROOT is None: raise RuntimeError("Open this notebook inside death-knight-lab.")
if str(ROOT) not in sys.path: sys.path.insert(0, str(ROOT))
from standard.catalog import load_catalog, load_classes, coverage_report, show_overview, search_cards, show_cards

cards = load_catalog()
classes = load_classes()
report = coverage_report(cards)
print(f"Loaded {len(cards):,} cards and {len(classes)} class definitions.")
print("No matches or training have started.")
```

```python
show_overview(cards, classes, report)
```

Change `HERO_CLASS` to a class name or `ALL`; neutral cards are included by default. `QUERY` searches card names, rules text, IDs, mechanics, and rune requirements. Set `MAX_ROWS = 1500` to display the entire book. Then rerun the following two cells.

## Rotation and release decisions
Included sets: Core, Into the Emerald Dream, The Lost City of Un’Goro, Across the Timeways, CATACLYSM, Escape from Violet Hold, and released Event cards. Mini-sets and class sets are included under their parent sets.

M.O.T.H.E.R. is included as an already-released card from the upcoming expansion. Other currently cataloged Black Empire cards and the three October 6 event cards are excluded from this date’s pool. Wild, cosmetic, trial-copy, Battlegrounds, and unreleased records are retained in a separate reference archive, **not** mixed into the deck pool. That reference archive is not a valid random-generation pool.

The date is frozen deliberately. Refreshing a future experiment requires reviewing rotation, release exceptions, balance changes, and bans again.

```python
manifest = json.loads((ROOT / "data" / "standard" / "manifest.json").read_text())
rules = json.loads((ROOT / "data" / "standard" / "release_rules.json").read_text())
print("As of:", manifest["as_of"])
print("Patch:", manifest["patch"])
print("Source:", manifest["source"])
print("Legality:", manifest["legality_status"])
print("Excluded future cards:")
for card_id, release_date in rules["excluded_future_cards"].items(): print(" ", card_id, release_date)
```

## Save the exact coverage report
This lists every catalog card and whether its data matches an explicitly implemented effect in the old Death Knight engine. An effect implemented in that small environment is not yet validated across all Standard interactions. Nothing is silently treated as a plain minion or a spell with no effect.

```python
output = ROOT / "runs" / "standard_catalog_2026-09-18"
output.mkdir(parents=True, exist_ok=True)
report_path = output / "coverage.json"
temporary = report_path.with_suffix(".tmp")
temporary.write_text(json.dumps(report, indent=2) + "\n")
temporary.replace(report_path)
print("Saved:", report_path)
print("FULL STANDARD SIMULATION READY:", report["full_standard_ready"])
```
